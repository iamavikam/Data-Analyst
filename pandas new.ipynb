# Pandas Attributes: Examples of Every Attribute
Basics to advanced. Attributes are accessed **without parentheses** (`df.shape`), unlike methods (`df.head()`).

```python
import pandas as pd
import numpy as np

print("pandas version:", pd.__version__)

df = pd.DataFrame({
    "name":   ["Asha", "Ravi", "Meena", "Karan", "Sita", "Dev"],
    "age":    [25, 32, 28, 45, 38, 29],
    "city":   ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai", "Delhi"],
    "salary": [50000.0, 72000.5, 61000.0, 95000.0, 83000.0, 58000.0],
    "joined": pd.to_datetime(["2020-01-15", "2018-06-01", "2021-03-20",
                              "2015-11-30", "2019-08-10", "2022-02-28"]),
})
df
```

## 1. Basic Attributes
| Attribute | Meaning |
|---|---|
| `shape` | (rows, columns) |
| `size` | total elements |
| `ndim` | number of dimensions |
| `columns` | column labels |
| `index` | row labels |
| `dtypes` | type of each column |
| `values` | data as NumPy array |
| `empty` | True if no data |
| `T` | transpose |
| `axes` | [index, columns] |
| `name` (Series) | name of a Series |

```python
print("shape  :", df.shape)
print("size   :", df.size)
print("ndim   :", df.ndim)
print("columns:", df.columns)
print("index  :", df.index)
print("empty  :", df.empty, "| empty DataFrame ->", pd.DataFrame().empty)
```

```python
print("dtypes:\n", df.dtypes)
```

```python
print("values:\n", df.values)
```

```python
print("T (transpose):")
df.T
```

```python
print("axes:", df.axes)

s = df["age"]
print("Series name:", s.name)
s.name = "age_in_years"
print("Renamed    :", s.name)
print("Series ndim:", s.ndim, "| shape:", s.shape, "| size:", s.size)
```

## 2. Indexing Attributes
| Attribute | Meaning |
|---|---|
| `loc[]` | label-based selection |
| `iloc[]` | position-based selection |
| `at[]` | fast single value by label |
| `iat[]` | fast single value by position |

```python
# loc: label based (end of slice is inclusive)
print(df.loc[0, "name"])
print(df.loc[1:3, ["name", "city"]])
print(df.loc[df["age"] > 30, ["name", "age"]])  # boolean mask
```

```python
# iloc: position based (end of slice is exclusive)
print(df.iloc[0, 1])
print(df.iloc[1:3, [0, 2]])
print(df.iloc[-1])  # last row
```

```python
# at / iat: fastest way to get or set ONE value
print("at  :", df.at[2, "name"])
print("iat :", df.iat[2, 0])

df2 = df.copy()
df2.at[0, "salary"] = 55000.0
df2.iat[1, 1] = 33
df2.head(2)
```

## 3. Memory & Type Info
`dtypes`, `memory_usage()` (method, used with attributes), `flags`, `attrs`.

```python
print(df.dtypes)
print()
print(df.memory_usage())
print("deep:", df.memory_usage(deep=True).sum(), "bytes")
```

```python
# flags: behaviour flags of the object
print(df.flags)
print("allows_duplicate_labels:", df.flags.allows_duplicate_labels)

d = pd.DataFrame({"a": [1, 2]}, index=["x", "y"])
d.flags.allows_duplicate_labels = False
try:
    d.reindex(["x", "x"])
except Exception as e:
    print(type(e).__name__, "->", e)
```

```python
# attrs: dictionary for custom metadata
df.attrs["source"] = "employees.csv"
df.attrs["owner"]  = "HR"
print(df.attrs)
```

## 4. Index Attributes (`df.index.<attr>`)

```python
idx = df.index
print("is_unique               :", idx.is_unique)
print("is_monotonic_increasing :", idx.is_monotonic_increasing)
print("is_monotonic_decreasing :", idx.is_monotonic_decreasing)
print("hasnans                 :", idx.hasnans)
print("names                   :", idx.names)
print("nlevels                 :", idx.nlevels)
print("inferred_type           :", idx.inferred_type)
print("dtype                   :", idx.dtype)
print("shape / size / ndim     :", idx.shape, idx.size, idx.ndim)
print("values                  :", idx.values)
```

```python
# Duplicate / unsorted index
bad = pd.Index([3, 1, 1, 2])
print("is_unique:", bad.is_unique)
print("monotonic increasing:", bad.is_monotonic_increasing)

# Named index
named = df.set_index("name")
print("names:", named.index.names)
```

```python
# MultiIndex: levels, codes, nlevels, names
mi = df.set_index(["city", "name"]).sort_index().index
print(mi)
print("nlevels:", mi.nlevels)
print("names  :", mi.names)
print("levels :", mi.levels)
print("codes  :", mi.codes)
```

## 5. Series-Specific Attributes

```python
s = df["salary"]
print("dtype                    :", s.dtype)
print("is_unique                :", s.is_unique)
print("hasnans                  :", s.hasnans)
print("is_monotonic_increasing  :", s.is_monotonic_increasing)
print("is_monotonic_decreasing  :", s.is_monotonic_decreasing)
print("nbytes                   :", s.nbytes)
print("array                    :", s.array)
print("values                   :", s.values)
print("index                    :", s.index)
print("name                     :", s.name)
```

```python
s_nan = pd.Series([1, 2, np.nan, 2])
print("hasnans  :", s_nan.hasnans)
print("is_unique:", s_nan.is_unique)

s_sorted = pd.Series([1, 2, 3, 4])
print("sorted increasing:", s_sorted.is_monotonic_increasing)
```

## 6. Accessors (Intermediate to Advanced)
Namespaces that unlock specialised tools: `.str`, `.dt`, `.cat`, `.sparse`, `.plot`, `.style`.

### 6.1 `.str` (text)

```python
names = df["name"]
print(names.str.upper().tolist())
print(names.str.len().tolist())
print(names.str.contains("a").tolist())
print(names.str.startswith("R").tolist())
print(names.str.replace("a", "@").tolist())
print(pd.Series(["a-b", "c-d"]).str.split("-").tolist())
```

### 6.2 `.dt` (datetime): every common attribute

```python
d = df["joined"]
dt_attrs = ["year", "month", "day", "hour", "minute", "second",
            "dayofweek", "dayofyear", "quarter", "days_in_month",
            "is_month_start", "is_month_end", "is_quarter_end",
            "is_year_end", "is_leap_year"]
out = pd.DataFrame({a: getattr(d.dt, a) for a in dt_attrs})
out.insert(0, "joined", d)
out
```

```python
print("date:\n", d.dt.date.head(2).tolist())
print("time:\n", d.dt.time.head(2).tolist())
print("tz  :", d.dt.tz)
print("day_name():", d.dt.day_name().tolist())
print("month_name():", d.dt.month_name().tolist())

# Timezone attribute after localising
d_utc = d.dt.tz_localize("UTC")
print("tz after tz_localize:", d_utc.dt.tz)
```

### 6.3 `.cat` (categorical)

```python
c = df["city"].astype("category")
print("categories :", c.cat.categories)
print("codes      :", c.cat.codes.tolist())
print("ordered    :", c.cat.ordered)

# Ordered categorical
sizes = pd.Series(["S", "L", "M"]).astype(
    pd.CategoricalDtype(["S", "M", "L"], ordered=True))
print("ordered    :", sizes.cat.ordered)
print("max        :", sizes.max())
```

### 6.4 `.sparse`

```python
sp = pd.Series(pd.arrays.SparseArray([0, 0, 1, 0, 0, 2, 0, 0]))
print("dtype    :", sp.dtype)
print("density  :", sp.sparse.density)
print("fill_value:", sp.sparse.fill_value)
print("npoints  :", sp.sparse.npoints)
print("sp_values:", sp.sparse.sp_values)
```

### 6.5 `.plot`

```python
try:
    df.plot.bar(x="name", y="salary", title="Salary by person")
    df.plot.line(x="joined", y="salary", marker="o")
    df.plot.scatter(x="age", y="salary")
    df["city"].value_counts().plot.pie()
    df["salary"].plot.hist(bins=4)
    df["salary"].plot.box()
    print("plots created")
except ImportError as e:
    print("matplotlib not installed:", e)
```

### 6.6 `.style` (HTML styling)

```python
try:
    display(df[["name", "age", "salary"]].style
            .highlight_max(subset=["salary"], color="lightgreen")
            .highlight_min(subset=["salary"], color="salmon")
            .format({"salary": "{:,.0f}"}))
except ImportError as e:
    print("Styler needs jinja2:", e)
```

## 7. Groupby, Rolling & Resample Objects

```python
g = df.groupby("city")
print("ngroups :", g.ngroups)
print("groups  :", g.groups)
print("indices :", g.indices)
print("size()  :\n", g.size())
```

```python
# Rolling window attributes
r = df["salary"].rolling(3)
print("window :", r.window)
print("min_periods:", r.min_periods)
print("center :", r.center)
print(r.mean().tolist())
```

```python
# Resample (time-based grouping)
ts = df.set_index("joined")["salary"]
rs = ts.resample("YE")  # use "Y" on pandas < 2.2
print(type(rs).__name__)
print(rs.sum())
```

## 8. Global Options (`pd.options`)

```python
print("max_rows        :", pd.options.display.max_rows)
print("max_columns     :", pd.options.display.max_columns)
print("float_format    :", pd.options.display.float_format)
print("precision       :", pd.options.display.precision)
print("width           :", pd.options.display.width)
```

```python
# Change options temporarily with option_context
with pd.option_context("display.max_rows", 2, "display.precision", 1):
    print(df[["name", "salary"]])

# Change permanently for the session
pd.options.display.max_columns = None   # show all columns
pd.options.display.float_format = "{:,.2f}".format
df[["name", "salary"]].head(3)
```

```python
# Copy-on-Write and chained assignment
print("copy_on_write       :", pd.options.mode.copy_on_write)
print("chained_assignment  :", pd.options.mode.chained_assignment)

# Reset the options we changed
pd.reset_option("display.float_format")
pd.reset_option("display.max_columns")
```
