# Matplotlib Attributes: Examples of Every Attribute

```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

print("Matplotlib version:", mpl.__version__)

x = np.linspace(0, 10, 50)
y = np.sin(x)
```

## 1. Module-Level Attributes
| Attribute | Meaning |
|---|---|
| `mpl.__version__` | installed version |
| `mpl.rcParams` | current default settings (dict-like) |
| `mpl.rcParamsDefault` | factory defaults |
| `mpl.get_backend()` | active rendering backend |
| `plt.style.available` | built-in style names |
| `mpl.colormaps` | registry of colormaps |

```python
print("version        :", mpl.__version__)
print("backend        :", mpl.get_backend())
print("rcParams size  :", len(mpl.rcParams))
print("config dir     :", mpl.get_configdir())
print("matplotlibrc   :", mpl.matplotlib_fname())
print("styles         :", plt.style.available[:6], "...", len(plt.style.available), "total")
print("colormaps      :", len(list(mpl.colormaps)), "total, e.g.", list(mpl.colormaps)[:5])
```

## 2. `Figure` Attributes
| Attribute | Meaning |
|---|---|
| `fig.axes` | list of Axes in the figure |
| `fig.dpi` | dots per inch |
| `fig.figsize` / `get_size_inches()` | size in inches |
| `fig.number` | figure number (pyplot) |
| `fig.canvas` | drawing canvas |
| `fig.patch` | background rectangle |
| `fig.bbox` | figure box in pixels |
| `fig.subplotpars` | subplot margins |
| `fig.texts` / `fig.lines` / `fig.patches` | figure-level artists |
| `fig.legends` | figure-level legends |

```python
fig, ax = plt.subplots(figsize=(6, 4), dpi=100)
ax.plot(x, y, label="sin(x)")
fig.suptitle("My Figure")

print("fig.axes          :", fig.axes)
print("number of axes    :", len(fig.axes))
print("fig.dpi           :", fig.dpi)
print("size (inches)     :", fig.get_size_inches())
print("fig.number        :", fig.number)
print("canvas            :", type(fig.canvas).__name__)
print("canvas size (px)  :", fig.canvas.get_width_height())
print("bbox (px)         :", fig.bbox.width, "x", fig.bbox.height)
print("patch facecolor   :", fig.patch.get_facecolor())
print("suptitle text     :", fig.get_suptitle())
print("texts / lines     :", len(fig.texts), len(fig.lines))
print("layout engine     :", fig.get_layout_engine())
```

```python
sp = fig.subplotpars
print("subplotpars -> left={:.3f} right={:.3f} bottom={:.3f} top={:.3f} wspace={:.2f} hspace={:.2f}".format(
    sp.left, sp.right, sp.bottom, sp.top, sp.wspace, sp.hspace))
print("tight / constrained:", fig.get_tight_layout(), fig.get_constrained_layout())
print("figure label       :", repr(fig.get_label()))
print("all fig numbers    :", plt.get_fignums())
```

## 3. `Axes` Attributes
| Attribute | Meaning |
|---|---|
| `ax.figure` | parent figure |
| `ax.xaxis` / `ax.yaxis` | the two Axis objects |
| `ax.spines` | the four borders |
| `ax.title` | center title (Text) |
| `ax.patch` | background rectangle |
| `ax.bbox` | Axes box in pixels |
| `ax.dataLim` | bounding box of the data |
| `ax.viewLim` | currently visible limits |
| `ax.transData` / `ax.transAxes` | coordinate transforms |
| `ax.legend_` | legend object (or None) |
| `ax.name` | projection name |

```python
fig, ax = plt.subplots()
ax.plot(x, y, label="sin")
ax.set_title("Sine wave")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()

print("ax.figure is fig  :", ax.figure is fig)
print("ax.name           :", ax.name)
print("ax.title text     :", ax.title.get_text())
print("ax.xaxis          :", ax.xaxis)
print("ax.yaxis          :", ax.yaxis)
print("ax.spines keys    :", list(ax.spines))
print("ax.legend_        :", ax.legend_)
print("ax.patch color    :", ax.patch.get_facecolor())
print("ax.bbox (px)      :", ax.bbox.bounds)
print("ax.dataLim bounds :", ax.dataLim.bounds)
print("ax.viewLim bounds :", ax.viewLim.bounds)
print("ax.get_xlim()     :", ax.get_xlim())
print("ax.get_ylim()     :", ax.get_ylim())
print("ax.get_aspect()   :", ax.get_aspect())
```

```python
# Other title positions and axes-level text
ax.set_title("Left", loc="left")
ax.set_title("Right", loc="right")
print("left title  :", ax.get_title(loc="left"))
print("right title :", ax.get_title(loc="right"))
print("xlabel      :", ax.xaxis.label.get_text())
print("ylabel      :", ax.yaxis.label.get_text())
print("xscale      :", ax.get_xscale(), "| yscale:", ax.get_yscale())
print("grid visible:", ax.xaxis.get_gridlines()[0].get_visible())
```

## 4. Artist Lists on an Axes
Everything drawn on an Axes is an **Artist**, stored in these lists.

| Attribute | Contains |
|---|---|
| `ax.lines` | `Line2D` from `plot` |
| `ax.patches` | `Rectangle` / `Wedge` etc. from `bar`, `hist`, `pie` |
| `ax.texts` | `Text` from `ax.text` / `annotate` |
| `ax.collections` | `PathCollection` from `scatter`, `fill_between` |
| `ax.images` | `AxesImage` from `imshow` |
| `ax.containers` | `BarContainer`, `ErrorbarContainer`, `StemContainer` |
| `ax.artists` | misc artists |
| `ax.child_axes` | inset axes |

```python
fig, ax = plt.subplots()
ax.plot(x, y, "r--", label="line")
ax.scatter(x[::5], y[::5], color="blue")
ax.bar([1, 2, 3], [0.2, 0.5, 0.3])
ax.text(5, 0.5, "hello")
ax.annotate("peak", xy=(1.5, 1), xytext=(3, 1.1), arrowprops=dict(arrowstyle="->"))
ax.fill_between(x, y, alpha=0.2)

print("lines       :", ax.lines)
print("patches     :", len(ax.patches), "->", type(ax.patches[0]).__name__)
print("texts       :", [t.get_text() for t in ax.texts])
print("collections :", [type(c).__name__ for c in ax.collections])
print("containers  :", ax.containers)
print("images      :", ax.images)
print("child_axes  :", ax.child_axes)
print("children    :", len(ax.get_children()), "artists total")
```

## 5. `Axis` Attributes and Ticks
`ax.xaxis` and `ax.yaxis` hold tick and label information.

| Attribute | Meaning |
|---|---|
| `axis.label` | axis label (Text) |
| `axis.majorTicks` / `minorTicks` | tick objects |
| `axis.major` / `axis.minor` | tick locator/formatter holders |
| `axis.major.locator` / `.formatter` | tick positions / labels |
| `axis.offsetText` | scientific offset text |
| `axis.axis_name` | `'x'` or `'y'` |
| `axis.units` | unit converter |

```python
fig, ax = plt.subplots()
ax.plot(x, y)
fig.canvas.draw()      # ticks are created at draw time

xa, ya = ax.xaxis, ax.yaxis
print("axis_name        :", xa.axis_name, ya.axis_name)
print("label            :", repr(xa.label.get_text()))
print("major locator    :", xa.major.locator)
print("major formatter  :", xa.major.formatter)
print("minor locator    :", xa.minor.locator)
print("n major ticks    :", len(xa.majorTicks))
print("tick locations   :", xa.get_majorticklocs())
print("tick labels      :", [t.get_text() for t in xa.get_majorticklabels()])
print("offsetText       :", repr(xa.offsetText.get_text()))
print("units            :", xa.units)
print("pickradius       :", xa.get_pickradius())
```

```python
# Tick object attributes
t = ax.xaxis.majorTicks[0]
print("tick1line :", t.tick1line)
print("tick2line :", t.tick2line)
print("gridline  :", t.gridline)
print("label1    :", t.label1)
print("tick loc  :", t.get_loc())
print("pad       :", t.get_pad())

# Customize via locator/formatter
from matplotlib.ticker import MultipleLocator, FormatStrFormatter
ax.xaxis.set_major_locator(MultipleLocator(2))
ax.xaxis.set_major_formatter(FormatStrFormatter("%.1f"))
ax.xaxis.set_minor_locator(MultipleLocator(0.5))
fig.canvas.draw()
print("\nnew major locs:", ax.xaxis.get_majorticklocs())
print("new minor count:", len(ax.xaxis.get_minorticklocs()))
```

## 6. Spines
`ax.spines` is a dict-like of the four borders: `left`, `right`, `top`, `bottom`.

```python
fig, ax = plt.subplots()
ax.plot(x, y)
for name, sp in ax.spines.items():
    print(f"{name:7} visible={sp.get_visible()} linewidth={sp.get_linewidth()} "
          f"color={sp.get_edgecolor()} position={sp.get_position()} type={sp.spine_type}")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_position(("data", 5))
ax.spines["bottom"].set_position("zero")
print("\nleft position now:", ax.spines["left"].get_position())
```

## 7. Common `Artist` Attributes
Every drawn object inherits these.

| Attribute | Meaning |
|---|---|
| `axes` / `figure` | parent Axes / Figure |
| `zorder` | drawing order |
| `alpha` | transparency |
| `visible` | shown or hidden |
| `label` | legend label |
| `gid` | group id |
| `stale` | needs redraw? |
| `clip_on` / `clip_box` | clipping |
| `transform` | coordinate transform |
| `animated` | blitting flag |
| `picker` | pick-event setting |

```python
fig, ax = plt.subplots()
(line,) = ax.plot(x, y, label="sin", alpha=0.7, zorder=3, gid="my-line")

print("axes is ax       :", line.axes is ax)
print("figure is fig    :", line.figure is fig)
print("zorder           :", line.zorder)
print("alpha            :", line.get_alpha())
print("visible          :", line.get_visible())
print("label            :", line.get_label())
print("gid              :", line.get_gid())
print("clip_on          :", line.get_clip_on())
print("animated         :", line.get_animated())
print("picker           :", line.get_picker())
print("transform        :", type(line.get_transform()).__name__)
print("stale            :", line.stale)
print("mouseover        :", line.mouseover)
print("sketch params    :", line.get_sketch_params())
```

```python
# Artists are mutable: change attributes then redraw
line.set_visible(False)
print("visible now:", line.get_visible())
line.set_visible(True)

# Inspect all settable properties
import matplotlib.artist as martist
props = martist.getp(line)   # prints all properties
```

## 8. `Line2D`, `Rectangle`, `Text` Attributes

```python
fig, ax = plt.subplots()
(ln,) = ax.plot([1, 2, 3], [4, 5, 6], color="green", linestyle="--", linewidth=2,
                marker="o", markersize=8, markerfacecolor="yellow", markeredgecolor="black")

print("Line2D")
print("  xdata/ydata    :", ln.get_xdata(), ln.get_ydata())
print("  xydata (N,2)   :\n", ln.get_xydata())
print("  color          :", ln.get_color())
print("  linestyle      :", ln.get_linestyle())
print("  linewidth      :", ln.get_linewidth())
print("  marker         :", ln.get_marker())
print("  markersize     :", ln.get_markersize())
print("  mfc / mec      :", ln.get_markerfacecolor(), ln.get_markeredgecolor())
print("  drawstyle      :", ln.get_drawstyle())
print("  path vertices  :", ln.get_path().vertices.shape)
```

```python
bars = ax.bar([1, 2, 3], [3, 5, 2], width=0.5, color="orange", edgecolor="k")
rect = bars.patches[0]
print("BarContainer")
print("  patches        :", len(bars.patches))
print("  datavalues     :", bars.datavalues)
print("  errorbar       :", bars.errorbar)
print("Rectangle")
print("  xy             :", rect.get_xy())
print("  width / height :", rect.get_width(), rect.get_height())
print("  x / y          :", rect.get_x(), rect.get_y())
print("  facecolor      :", rect.get_facecolor())
print("  edgecolor      :", rect.get_edgecolor())
print("  linewidth      :", rect.get_linewidth())
print("  bounds         :", rect.get_bbox().bounds)
```

```python
txt = ax.text(2, 4, "Hello", fontsize=14, color="red", ha="center", va="bottom",
              rotation=15, fontweight="bold", fontstyle="italic")
print("Text")
print("  text           :", txt.get_text())
print("  position       :", txt.get_position())
print("  fontsize       :", txt.get_fontsize())
print("  color          :", txt.get_color())
print("  rotation       :", txt.get_rotation())
print("  ha / va        :", txt.get_ha(), txt.get_va())
print("  fontweight     :", txt.get_fontweight())
print("  fontstyle      :", txt.get_fontstyle())
print("  fontfamily     :", txt.get_fontfamily())
print("  wrap           :", txt.get_wrap())
```

## 9. Legend Attributes

```python
fig, ax = plt.subplots()
ax.plot(x, y, label="sin")
ax.plot(x, np.cos(x), label="cos")
leg = ax.legend(title="Functions", loc="upper right", ncols=2, frameon=True)

print("legend           :", leg)
print("title            :", leg.get_title().get_text())
print("texts            :", [t.get_text() for t in leg.get_texts()])
handles = leg.legend_handles if hasattr(leg, "legend_handles") else leg.legendHandles
print("handles          :", handles)
print("frame on         :", leg.get_frame_on())
print("frame facecolor  :", leg.get_frame().get_facecolor())
print("bbox (px)        :", leg.get_window_extent(fig.canvas.get_renderer()).bounds if hasattr(fig.canvas, "get_renderer") else "n/a")

h, l = ax.get_legend_handles_labels()
print("handles/labels   :", h, l)
```

## 11. Transforms
Convert between coordinate systems.

| Transform | Coordinates |
|---|---|
| `ax.transData` | data coordinates |
| `ax.transAxes` | (0,0) bottom-left to (1,1) top-right of the Axes |
| `fig.transFigure` | (0,0) to (1,1) of the Figure |
| `fig.dpi_scale_trans` | inches |
| `ax.get_xaxis_transform()` | x in data, y in axes |

```python
fig, ax = plt.subplots(figsize=(6, 4), dpi=100)
ax.set_xlim(0, 10); ax.set_ylim(0, 1)
fig.canvas.draw()

print("data (5, 0.5)   -> pixels :", ax.transData.transform((5, 0.5)))
print("axes (0.5, 0.5) -> pixels :", ax.transAxes.transform((0.5, 0.5)))
print("fig  (0.5, 0.5) -> pixels :", fig.transFigure.transform((0.5, 0.5)))
print("pixels -> data            :", ax.transData.inverted().transform((300, 200)))
print("inches -> pixels (1, 1)   :", fig.dpi_scale_trans.transform((1, 1)))

# Place text using different coordinate systems
ax.text(0.5, 0.9, "axes coords", transform=ax.transAxes, ha="center")
ax.text(5, 0.2, "data coords", transform=ax.transData, ha="center")
```

## 12. Subplots & GridSpec

```python
fig, axs = plt.subplots(2, 3, figsize=(9, 5))
print("axs type    :", type(axs).__name__, "| shape:", axs.shape)
print("axs.flat[0] :", axs.flat[0])
print("fig.axes    :", len(fig.axes))

ss = axs[0, 1].get_subplotspec()
print("rowspan     :", ss.rowspan, "| colspan:", ss.colspan)
print("num1        :", ss.num1)

gs = ss.get_gridspec()
print("GridSpec nrows/ncols:", gs.nrows, gs.ncols)
print("width_ratios        :", gs.get_width_ratios())
print("height_ratios       :", gs.get_height_ratios())
print("is_first_row/col    :", axs[0, 0].get_subplotspec().is_first_row(),
      axs[0, 0].get_subplotspec().is_first_col())
```

```python
# Custom grid with GridSpec
fig = plt.figure(figsize=(6, 5))
gs = fig.add_gridspec(2, 2, width_ratios=[2, 1], height_ratios=[1, 3])
a1 = fig.add_subplot(gs[0, :])
a2 = fig.add_subplot(gs[1, 0])
a3 = fig.add_subplot(gs[1, 1])
print("gs.nrows/ncols :", gs.nrows, gs.ncols)
print("width_ratios   :", gs.get_width_ratios())
print("a1 colspan     :", a1.get_subplotspec().colspan)

# Shared axes
fig, (s1, s2) = plt.subplots(2, 1, sharex=True)
print("\nshared x group:", s1.get_shared_x_axes().joined(s1, s2))
```

## 13. Images & Colorbars

```python
data = np.random.default_rng(0).random((5, 5))
fig, ax = plt.subplots()
im = ax.imshow(data, cmap="plasma", vmin=0, vmax=1, origin="upper", interpolation="nearest")
cb = fig.colorbar(im, ax=ax, label="value")

print("ax.images      :", ax.images)
print("im.cmap.name   :", im.cmap.name)
print("im.norm        :", type(im.norm).__name__, im.norm.vmin, im.norm.vmax)
print("im.get_array() shape:", im.get_array().shape)
print("im.origin      :", im.origin)
print("im.get_extent():", im.get_extent())
print("im.get_interpolation():", im.get_interpolation())

print("\nColorbar")
print("  cb.ax        :", cb.ax)
print("  cb.mappable  :", cb.mappable is im)
print("  cb.cmap.name :", cb.cmap.name)
print("  cb.norm      :", cb.norm)
print("  cb.vmin/vmax :", cb.vmin, cb.vmax)
print("  cb.orientation:", cb.orientation)
print("  cb.ax.get_ylabel():", cb.ax.get_ylabel())
print("  fig.axes     :", len(fig.axes), "(main + colorbar)")
```

```python
# Scatter exposes the same mappable attributes
fig, ax = plt.subplots()
sc = ax.scatter(x, y, c=y, cmap="coolwarm", s=40)
print("offsets shape :", sc.get_offsets().shape)
print("sizes[:3]     :", sc.get_sizes()[:3])
print("array[:3]     :", sc.get_array()[:3])
print("cmap          :", sc.get_cmap().name)
print("clim          :", sc.get_clim())
print("paths         :", len(sc.get_paths()))
```

## 14. `rcParams`

## Summary
- **Hierarchy:** `Figure` -> `Axes` -> `Axis`, `Spines`, artists (`lines`, `patches`, `texts`, `collections`, `images`, `containers`)
- **Figure:** `axes`, `dpi`, `canvas`, `patch`, `bbox`, `number`, `subplotpars`
- **Axes:** `figure`, `xaxis`, `yaxis`, `spines`, `title`, `legend_`, `dataLim`, `viewLim`, `transData`, `transAxes`
- **Axis:** `label`, `majorTicks`, `minorTicks`, `major.locator`, `major.formatter`, `offsetText`
- **Artist (common):** `zorder`, `alpha`, `visible`, `label`, `stale`, `transform`, `axes`, `figure`
- **Colors:** `BASE_COLORS`, `TABLEAU_COLORS`, `CSS4_COLORS`, `cmap.name`, `cmap.N`, `norm.vmin`, `norm.vmax`
- **Config:** `rcParams`, `rcParamsDefault`, `plt.style.available`, `get_backend()`
- **Lines/markers:** `Line2D.lineStyles`, `Line2D.markers`, `MarkerStyle`
